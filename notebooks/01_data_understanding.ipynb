```python
import pandas as pd
import numpy as np
import os
from pathlib import Path
```

```python
DATA_PATH = Path("../data/raw")

csv_files = list(DATA_PATH.glob("*.csv"))

print("Total CSV files:", len(csv_files))

for file in csv_files:
    print(f"{file.name} | {file.stat().st_size / (1024 * 1024):.2f} MB")
```

```python
for file in csv_files:
    df = pd.read_csv(file)

    print(f"\n{file.name}")
    print("Exact duplicate rows:", df.duplicated().sum())
```

```python
key_columns = {
    "olist_customers_dataset.csv": ["customer_id", "customer_unique_id"],
    "olist_orders_dataset.csv": ["order_id"],
    "olist_order_items_dataset.csv": ["order_id", "order_item_id"],
    "olist_order_payments_dataset.csv": ["order_id", "payment_sequential"],
    "olist_products_dataset.csv": ["product_id"],
    "olist_sellers_dataset.csv": ["seller_id"],
    "olist_order_reviews_dataset.csv": ["review_id"],
}

for filename, keys in key_columns.items():
    df = pd.read_csv(DATA_PATH / filename)

    print(f"\n{filename}")

    for key in keys:
        print(
            key,
            "Unique:", df[key].nunique(),
            "Duplicates:", df[key].duplicated().sum()
        )
```

```python
composite_keys = {
    "olist_order_items_dataset.csv": [
        "order_id",
        "order_item_id"
    ],
    "olist_order_payments_dataset.csv": [
        "order_id",
        "payment_sequential"
    ]
}

for filename, columns in composite_keys.items():

    df = pd.read_csv(DATA_PATH / filename)

    duplicate_count = df.duplicated(
        subset=columns
    ).sum()

    print(f"\n{filename}")
    print("Composite key:", columns)
    print("Duplicate combinations:", duplicate_count)
```

```python
reviews = pd.read_csv(
    DATA_PATH / "olist_order_reviews_dataset.csv"
)

duplicate_reviews = reviews[
    reviews.duplicated(
        subset=["review_id"],
        keep=False
    )
].sort_values("review_id")

print("Duplicate review records:", len(duplicate_reviews))

display(duplicate_reviews.head(20))

print("\nDuplicate review IDs across different orders:")
print(
    duplicate_reviews.groupby("review_id")["order_id"]
    .nunique()
    .value_counts()
)
```

```python
geo = pd.read_csv(
    DATA_PATH / "olist_geolocation_dataset.csv"
)

print("Original rows:", len(geo))

print(
    "Unique ZIP prefixes:",
    geo["geolocation_zip_code_prefix"].nunique()
)

print(
    "Unique coordinate combinations:",
    geo.drop_duplicates(
        subset=[
            "geolocation_zip_code_prefix",
            "geolocation_lat",
            "geolocation_lng"
        ]
    ).shape[0]
)

print("\nMost frequent ZIP prefixes:")

display(
    geo["geolocation_zip_code_prefix"]
    .value_counts()
    .head(10)
)
```

```python
reviews = pd.read_csv(
    DATA_PATH / "olist_order_reviews_dataset.csv"
)

print("Review ID + Order ID duplicates:",
      reviews.duplicated(
          subset=["review_id", "order_id"]
      ).sum())

print("Unique review IDs:", reviews["review_id"].nunique())
print("Unique orders:", reviews["order_id"].nunique())
```

```python
review_columns = [
    "review_score",
    "review_comment_title",
    "review_comment_message",
    "review_creation_date",
    "review_answer_timestamp"
]

duplicate_review_ids = reviews[
    reviews.duplicated("review_id", keep=False)
]

consistency = (
    duplicate_review_ids.groupby("review_id")[review_columns]
    .nunique(dropna=False)
)

print("Review IDs with consistent review information:",
      (consistency.max(axis=1) == 1).sum())

print("Review IDs with inconsistent review information:",
      (consistency.max(axis=1) > 1).sum())
```

```python
geo = pd.read_csv(
    DATA_PATH / "olist_geolocation_dataset.csv"
)

zip_summary = (
    geo.groupby("geolocation_zip_code_prefix")
    .agg(
        coordinate_count=("geolocation_lat", "count"),
        unique_latitudes=("geolocation_lat", "nunique"),
        unique_longitudes=("geolocation_lng", "nunique"),
        unique_cities=("geolocation_city", "nunique"),
        unique_states=("geolocation_state", "nunique")
    )
    .sort_values("coordinate_count", ascending=False)
)

display(zip_summary.head(10))
```

```python
customers = pd.read_csv(
    DATA_PATH / "olist_customers_dataset.csv",
    usecols=["customer_id", "customer_zip_code_prefix"]
)

orders = pd.read_csv(
    DATA_PATH / "olist_orders_dataset.csv",
    usecols=["order_id", "customer_id"]
)

order_items = pd.read_csv(
    DATA_PATH / "olist_order_items_dataset.csv",
    usecols=["order_id", "product_id", "seller_id"]
)

payments = pd.read_csv(
    DATA_PATH / "olist_order_payments_dataset.csv",
    usecols=["order_id"]
)

reviews = pd.read_csv(
    DATA_PATH / "olist_order_reviews_dataset.csv",
    usecols=["order_id"]
)

products = pd.read_csv(
    DATA_PATH / "olist_products_dataset.csv",
    usecols=["product_id", "product_category_name"]
)

sellers = pd.read_csv(
    DATA_PATH / "olist_sellers_dataset.csv",
    usecols=["seller_id", "seller_zip_code_prefix"]
)

geo = pd.read_csv(
    DATA_PATH / "olist_geolocation_dataset.csv",
    usecols=["geolocation_zip_code_prefix"]
)

category_translation = pd.read_csv(
    DATA_PATH / "product_category_name_translation.csv"
)

def check_foreign_key(child_df, child_column, parent_df, parent_column):

    parent_values = set(parent_df[parent_column].dropna())

    unmatched = child_df[
        child_df[child_column].notna() &
        ~child_df[child_column].isin(parent_values)
    ]

    print(f"\n{child_column} -> {parent_column}")
    print("Unmatched records:", len(unmatched))
    print("Unmatched unique values:", unmatched[child_column].nunique())


check_foreign_key(
    orders, "customer_id",
    customers, "customer_id"
)

check_foreign_key(
    order_items, "order_id",
    orders, "order_id"
)

check_foreign_key(
    order_items, "product_id",
    products, "product_id"
)

check_foreign_key(
    order_items, "seller_id",
    sellers, "seller_id"
)

check_foreign_key(
    payments, "order_id",
    orders, "order_id"
)

check_foreign_key(
    reviews, "order_id",
    orders, "order_id"
)

check_foreign_key(
    customers, "customer_zip_code_prefix",
    geo, "geolocation_zip_code_prefix"
)

check_foreign_key(
    sellers, "seller_zip_code_prefix",
    geo, "geolocation_zip_code_prefix"
)

check_foreign_key(
    products.dropna(subset=["product_category_name"]),
    "product_category_name",
    category_translation,
    "product_category_name"
)
```

```python
customer_zip_missing = customers[
    ~customers["customer_zip_code_prefix"].isin(
        geo["geolocation_zip_code_prefix"]
    )
]

seller_zip_missing = sellers[
    ~sellers["seller_zip_code_prefix"].isin(
        geo["geolocation_zip_code_prefix"]
    )
]

print("Unmatched customer ZIP prefixes:")
display(
    customer_zip_missing["customer_zip_code_prefix"]
    .value_counts()
    .head(20)
)

print("\nUnmatched seller ZIP prefixes:")
display(seller_zip_missing)
```

```python
unmatched_categories = products[
    products["product_category_name"].notna() &
    ~products["product_category_name"].isin(
        category_translation["product_category_name"]
    )
]

print("Unmatched category values:")
display(
    unmatched_categories["product_category_name"]
    .value_counts()
)

print("\nAffected product records:")
display(
    unmatched_categories[
        ["product_id", "product_category_name"]
    ].head(20)
)
```
